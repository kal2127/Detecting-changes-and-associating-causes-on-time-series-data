```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pymc as pm
import arviz as az

# 1. Load your dataset
# Assume your file is named 'brent_oil_prices.csv'
df = pd.read_csv('brent_oil_prices.csv')

# 2. Convert Date to a format Python understands
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# 3. Visualize the raw price to see trends and shocks
plt.figure(figsize=(12, 5))
plt.plot(df['Price'], label='Brent Price')
plt.title('Brent Oil Prices (1987-2022)')
plt.legend()
plt.show()

# 4. Create "Log Returns" to make the data more stable (stationary)
df['Log_Returns'] = np.log(df['Price']) - np.log(df['Price'].shift(1))
df.dropna(inplace=True) # Remove the first row which becomes empty
```
